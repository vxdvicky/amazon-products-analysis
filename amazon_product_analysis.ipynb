### 4.3 Data Type Conversion

Some numerical values in the dataset were stored as text because they contained currency symbols and commas. These values need to be converted into numerical format for accurate analysis.

```python
df["discount_price"] = (
    df["discount_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["discount_price"] = pd.to_numeric(df["discount_price"], errors="coerce")
```

```python
print(df["discount_price"].dtype)
```

```python
df[['discount_price']].head()
```

```python
df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["actual_price"] = pd.to_numeric(df["actual_price"], errors="coerce")
```

```python
df[['actual_price']].isnull().sum()
```

The `no_of_ratings` column contains commas and is converted into numerical format to enable quantitative analysis.

```python
df["no_of_ratings"] = (
    df["no_of_ratings"]
    .str.replace(",", "", regex=False)
)

df["no_of_ratings"] = pd.to_numeric(df["no_of_ratings"], errors="coerce")
```

```python
df["ratings"] = pd.to_numeric(df["ratings"], errors="coerce")
```

### 5.2 Discount Percentage

Discount percentage provides a relative measure of the discount offered on a product. It allows products with different price ranges to be compared fairly.



```python
df["discount_percentage"] = (
    (df["discount_amo"] / df["actual_price"]) * 100
).round(2)
```

```python
df[["actual_price", "discount_price", "discount_percentage"]].head()
```

### 5.3 Price Category

Products are categorized based on their discounted selling price to better understand the distribution of products across different price segments.

The categories are defined as:

- **Budget:** Less than ₹1,000
- **Mid Range:** ₹1,000 to ₹10,000
- **Premium:** Above ₹10,000

```python
def price_category(price):
    if price < 1000:
        return "Budget"
    elif price <= 10000:
        return "Mid Range"
    else:
        return "Premium"

df["price_category"] = df["discount_price"].apply(price_category)
```

```python
df[["discount_price", "price_category"]].head()
```

### 5.4 Rating Category

Products are categorized based on their customer ratings to simplify the analysis of customer satisfaction.

The categories are defined as:

- **Poor:** Rating below 3
- **Average:** Rating between 3 and 4
- **Excellent:** Rating above 4
- **No Rating:** No customer rating available

```python
def rating_category(rating):
    if pd.isna(rating):
        return "No Rating"
    elif rating < 3:
        return "Poor"
    elif rating <= 4:
        return "Average"
    else:
        return "Excellent"

df["rating_category"] = df["ratings"].apply(rating_category)
```
