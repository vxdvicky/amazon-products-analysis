### 4.3 Data Type Conversion

Some numerical values in the dataset were stored as text because they contained currency symbols and commas. These values need to be converted into numerical format for accurate analysis.

```python
df["discount_price"] = (
    df["discount_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["discount_price"] = pd.to_numeric(df["discount_price"], errors="coerce")
```

### 5.2 Discount Percentage

Discount percentage provides a relative measure of the discount offered on a product. It allows products with different price ranges to be compared fairly.



```python
df["discount_percentage"] = (
    (df["discount_amo"] / df["actual_price"]) * 100
).round(2)
```

### 5.3 Price Category

Products are categorized based on their discounted selling price to better understand the distribution of products across different price segments.

The categories are defined as:

- **Budget:** Less than ₹1,000
- **Mid Range:** ₹1,000 to ₹10,000
- **Premium:** Above ₹10,000

```python
def price_category(price):
    if price < 1000:
        return "Budget"
    elif price <= 10000:
        return "Mid Range"
    else:
        return "Premium"

df["price_category"] = df["discount_price"].apply(price_category)
```

```python
df[["discount_price", "price_category"]].head()
```

### 5.4 Rating Category

Products are categorized based on their customer ratings to simplify the analysis of customer satisfaction.

The categories are defined as:

- **Poor:** Rating below 3
- **Average:** Rating between 3 and 4
- **Excellent:** Rating above 4
- **No Rating:** No customer rating available

```python
def rating_category(rating):
    if pd.isna(rating):
        return "No Rating"
    elif rating < 3:
        return "Poor"
    elif rating <= 4:
        return "Average"
    else:
        return "Excellent"

df["rating_category"] = df["ratings"].apply(rating_category)
```

## 8. Business Recommendations

Based on the analysis of product categories, pricing, discounts, customer ratings, and customer rating activity, the following recommendations can be considered.

### 8.1 Focus on High-Engagement Categories

Categories with high customer rating activity can receive greater attention from sellers and businesses. The **TV, Audio & Cameras** category shows exceptionally high rating activity in the dataset.

However, total customer ratings should not be treated as direct sales or revenue because the dataset does not contain actual sales or revenue information.

### 8.2 Maintain Competitive Pricing

Budget products represent the largest price segment in the dataset. Businesses targeting a wider customer base can consider offering affordable products and competitive pricing.

Pricing strategies should be adjusted according to the category because customer preferences and average prices differ across product categories.

### 8.3 Use Discounts Carefully

The dataset shows a high average discount percentage. Businesses can use discounts and promotional pricing to attract customers, but discount strategies should be evaluated carefully to ensure that lower prices also support business profitability.

### 8.4 Consider Ratings Along with Rating Count

Product ratings should not be evaluated alone. A product with a high rating but very few customer ratings may provide less reliable evidence of customer satisfaction than a similarly rated product with a large number of ratings.

Therefore, businesses should consider both **average rating** and **number of customer ratings** when evaluating product performance.

### 8.5 Use Category-Specific Strategies

The analysis shows differences in pricing, discounts, and customer rating activity across categories. Businesses can therefore develop category-specific pricing and promotional strategies instead of applying the same strategy to all products.

### 8.6 Collect Sales and Revenue Data for Deeper Analysis

The current dataset provides product-level information such as prices, discounts, ratings, and customer rating counts. It does not provide actual sales, revenue, profit, or conversion data.

Adding these business metrics in future analysis would allow more accurate evaluation of product performance and profitability.
